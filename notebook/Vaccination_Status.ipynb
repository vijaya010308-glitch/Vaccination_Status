```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Excel file
df = pd.read_excel('/content/vaccination_status_clean.xlsx')

# Convert submission date to datetime
df['submitted_at'] = pd.to_datetime(
    df['submitted_at'], errors='coerce', utc=True
)

# Remove rows with missing required values
df = df.dropna(subset=['vaccination_status', 'submitted_at'])

sns.set_theme(style="whitegrid")

# 1. Vaccination Status Count
plt.figure(figsize=(9, 5))
sns.countplot(
    data=df,
    y='vaccination_status',
    order=df['vaccination_status'].value_counts().index
)
plt.title('Vaccination Status Distribution')
plt.xlabel('Number of Respondents')
plt.ylabel('Vaccination Status')
plt.tight_layout()
plt.show()

# 2. Vaccination Status Pie Chart
status_counts = df['vaccination_status'].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    status_counts,
    labels=status_counts.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Vaccination Status Percentage')
plt.tight_layout()
plt.show()

# 3. Submission Trend by Date
df['submission_date'] = df['submitted_at'].dt.date
daily = df.groupby('submission_date').size()

plt.figure(figsize=(10, 5))
daily.plot(kind='line', marker='o')
plt.title('Daily Vaccination Survey Submissions')
plt.xlabel('Submission Date')
plt.ylabel('Number of Submissions')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 4. Vaccination Status by Month
df['month'] = df['submitted_at'].dt.strftime('%Y-%m')

monthly = pd.crosstab(
    df['month'], df['vaccination_status']
)

monthly.plot(kind='bar', figsize=(10, 5))
plt.title('Monthly Vaccination Status Analysis')
plt.xlabel('Month')
plt.ylabel('Number of Respondents')
plt.xticks(rotation=45)
plt.legend(title='Vaccination Status', bbox_to_anchor=(1.02, 1))
plt.tight_layout()
plt.show()
```
